```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt

df = pd.read_csv('canada_per_capita_income.csv')
new_df = df.drop(df.columns[1], axis='columns')
c= df.loc[:, df.columns[1]]
reg = linear_model.LinearRegression()
reg.fit(new_df,c)
reg.predict([[2020]])
```

```python
import pickle
```

```python
with open('linear_model_pickle','wb') as file:
    pickle.dump(reg,file)
```
