```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
```

```python
df=pd.read_csv("art_restoration_dataset.csv")
```

```python
df.head()
```

```python
def restoration_level(x):
    if x<=5:
        return "Easy"
    elif x<=10:
        return "Moderate"
    elif x<=20:
        return "Difficult"
    else:
        return "Impossible"
df['restoration_level'] = df['damage_percent'].apply(restoration_level)
```

```python
def restoration_level(x):
    if x<=5:
        return 0
    elif x<=10:
        return 1
    elif x<=20:
        return 2
    else:
        return 3
df['restoration_level'] = df['damage_percent'].apply(restoration_level)
```

```python
df
```

```python
x=df[['damage_percent','blur_score','edge_count','brightness']]
y=df['restoration_level']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=GaussianNB()
model.fit(xtrain,ytrain)
```

```python
model.score(xtest,ytest)
```
