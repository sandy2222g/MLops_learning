```python
import cv2
import os

os.chdir(r"")

cap = cv2.VideoCapture(r"")

framerate = int(cap.get(cv2.CAP_PROP_FPS))

framecount = 0
count = 0

while True:
    success, frame = cap.read()

    if not success:
        break

    frame = cv2.resize(frame, (1280, 720))

    framecount += 1

    if framecount == (framerate * 1 // 4):
        print('hi', count)

        framecount = 0

        cv2.imwrite(f'118_frames{count}.jpg', frame)

        count += 1

    key = cv2.waitKey(1) & 0xFF

    if key == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
