```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("art_restoration_dataset.csv")
```

```python
def restoration_level(x):
    if x <= 15:
        return "Easy"
    elif x <= 30:
        return "Moderate"
    elif x <= 45:
        return "Difficult"
    else:
        return "Impossible"
df['restoration_level'] = df['damage_percent'].apply(restoration_level)
```

```python
df.head()
```

```python
x=df[['damage_percent','blur_score','edge_count','brightness']]
y=df['restoration_level']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=LogisticRegression(max_iter=1000)
model.fit(xtrain,ytrain)
```

```python
model.score(xtest,ytest)
```

```python
prediction = model.predict([[20, 1500, 4000, 120]])
```
